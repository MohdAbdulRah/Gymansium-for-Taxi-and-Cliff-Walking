```python
import gymnasium 
import random 
import numpy as np
```

```python
env = gymnasium.make("CliffWalking-v0",render_mode="rgb_array")
```

```python
print(env.action_space.n)
```

```python
starting_state,_ = env.reset() #Start State
starting_state
```

```python
alpha = 0.5
gamma = 0.99
episodes = 500
epsilon = 0.1
```

```python
Q = np.zeros((48,4))
```

```python
def greedy_epsilon(state):
  if(random.random() < epsilon):
    return env.action_space.sample()
  else:
    return np.argmax(Q[state])
```

```python
for episode in range(episodes):
  env = gymnasium.make("CliffWalking-v0")
  done = False

  state,_ = env.reset()
  action = greedy_epsilon(state)

  total_reward = 0
  episode_len = 0

  while not done:
    next_state,reward,terminated,truncate,_ = env.step(action)
    done = terminated or truncate
    next_action = greedy_epsilon(next_state)

    total_reward += reward
    episode_len += 1

    Q[state,action] = Q[state,action] + alpha*(reward + gamma*Q[next_state,next_action] - Q[state,action])
    state = next_state
    action = next_action
  env.close()
  print(f"Episode: {episode+1}, Total Reward: {total_reward}, Episode Length: {episode_len}")
         
```

```python
env = gymnasium.make("CliffWalking-v0",render_mode="human")
done = False
total_reward = 0
episode_len = 0
state,_ = env.reset()
action = np.argmax(Q[state])
while not done:
  next_state,reward,terminated,truncate,_ = env.step(action)
  next_action = np.argmax(Q[next_state])

  done = terminated or truncate

  total_reward += reward
  episode_len += 1

  state = next_state
  action = next_action
env.close()
```
