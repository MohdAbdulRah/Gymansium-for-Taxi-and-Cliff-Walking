```python
import gymnasium
```

```python
import numpy as np 
import random 
```

```python
env = gymnasium.make("Taxi-v3")
```

```python
env.observation_space.n
```

```python
env.action_space.n
```

```python
Q = np.zeros((500,6))
```

```python
alpha = 0.5 
gamma = 0.99 
epsilon = 0.1 

def epsilon_greedy(state):
    if(random.random() < epsilon):
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
```

```python
episodes = 5000

for episode in range(episodes):
    env = gymnasium.make("Taxi-v3")
    done = False 
    state,_ = env.reset()
    action = epsilon_greedy(state)
    total_reward = 0   
    episode_len = 0 

    while not done:
        next_state,reward,terminated,truncate,_ = env.step(action) 
        done = terminated or truncate 

        total_reward += reward
        episode_len += 1 

        next_action = epsilon_greedy(next_state)

        Q[state,action] = Q[state,action] + alpha*(reward + gamma*Q[next_state,next_action]- Q[state,action])

        state = next_state
        action = next_action
    env.close()
    print(f"Episode : {episode+1}/500 Reward : {total_reward}  Episode Len : {episode_len}")
```

```python
env = gymnasium.make("Taxi-v3",render_mode="human")
state,_ = env.reset() 
done=False
action = np.argmax(Q[state])
total_reward = 0
episode_len = 0

while not done:
    next_state,reward,terminated,truncate,_ = env.step(action)
    total_reward += reward
    episode_len += 1 
    done = terminated or truncate 
    next_action = np.argmax(Q[next_state])

    state = next_state
    action = next_action
    import time
    time.sleep(0.3)

env.close()
```
